```python
from pocket_tts import TTSModel
tts_model=TTSModel.load_model()

```

```python
voice_state=tts_model.get_state_for_audio_prompt(
    'alba'
)
```

```python
audio = tts_model.generate_audio(voice_state, "Hello world, this is a test.")
```

```python
import scipy.io.wavfile
```

```python
scipy.io.wavfile.write("output.wav", tts_model.sample_rate, audio.numpy())
```

```python
!pip install torchcodec
import torchaudio
import torch

# NOTE: Replace 'my_voice.wav' with the path to your audio file
# You can upload an audio file to your Colab environment by clicking the folder icon on the left pane.
waveform, sample_rate = torchaudio.load('/content/output.wav')

# Ensure the waveform is mono if it's stereo, or convert to a compatible format if needed
# Assuming pocket-tts expects a single channel, let's take the first channel if stereo
if waveform.shape[0] > 1:
    waveform = waveform[0].unsqueeze(0);

# Some models might require the waveform to be at a specific sample rate.
# If your audio's sample_rate is different from tts_model.sample_rate,
# you might need to resample it.
# For simplicity, we'll assume the loaded sample_rate is compatible or close enough for now.

# Get the voice state from your audio prompt
my_voice_state = tts_model.get_state_for_audio_prompt(waveform.cpu())

# Generate audio using your custom voice state
custom_audio = tts_model.generate_audio(my_voice_state, "This is my custom voice.")

# Save the generated audio
scipy.io.wavfile.write("my_custom_voice_output.wav", tts_model.sample_rate, custom_audio.numpy())
```
